```python
import os
import numpy as np
import pandas as pd
from scipy.stats import rankdata, wilcoxon
```

```python
# ============================================================
# 2. Load dataset
# ============================================================
FILE_NAME = "paired_list"

# Read the excel file
paired = pd.read_excel("../../"+FILE_NAME+".xlsx")

# Show the head
print(paired.tail())
```

```python
left = paired["pederson_left"].to_numpy()
right = paired["pederson_right"].to_numpy()

# Paired differences
diff = left - right

# Wilcoxon'daki gibi sıfır farkları çıkar
diff_nonzero = diff[diff != 0]

# Absolute differences için rank
ranks = rankdata(np.abs(diff_nonzero), method="average")

# Positive and negative rank sums
W_plus = ranks[diff_nonzero > 0].sum()
W_minus = ranks[diff_nonzero < 0].sum()

# Rank-biserial correlation
r_rb = (W_plus - W_minus) / (W_plus + W_minus)

print("W+ =", W_plus)
print("W- =", W_minus)
print("Rank-biserial correlation =", r_rb)

# Wilcoxon test
result = wilcoxon(
    left,
    right,
    alternative="two-sided",
    zero_method="wilcox"
)

print("Wilcoxon statistic =", result.statistic)
print("p-value =", result.pvalue)
```
